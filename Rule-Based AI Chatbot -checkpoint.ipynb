```python
"""
================================================================================
Module       : core.chatbot
Description  : Enterprise-Grade Rule-Based Conversational Engine
Author       : AI Engineering Intern
Architecture : Declarative Pattern Matching & Intent Mapping Engine
================================================================================
"""

import logging
import random
import re
from dataclasses import dataclass, field
from datetime import datetime
from typing import Callable, List

# Configure production logger
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(name)s: %(message)s"
)
logger = logging.getLogger("RuleBasedChatbot")

@dataclass
class UserSession:
    """Tracks state and session metrics for an active chat user."""
    user_name: str = "Guest"
    session_start: datetime = field(default_factory=datetime.now)
    interaction_count: int = 0

    def increment_interaction(self) -> None:
        self.interaction_count += 1


@dataclass
class IntentRule:
    """Defines a declarative intent rule for the chatbot decision matrix."""
    intent_id: str
    patterns: List[str]
    handler: Callable[[str, UserSession], str]
    is_regex: bool = False


# ==============================================================================
# BUSINESS LOGIC & HANDLERS
# ==============================================================================

class ResponseHandlers:
    """Encapsulates handler functions invoked by matched intents."""

    @staticmethod
    def handle_greeting(input_text: str, session: UserSession) -> str:
        greetings = [
            f"Hello {session.user_name}! How can I assist you today?",
            f"Greetings, {session.user_name}! What can I help you with?",
            f"Assalamualaikum {session.user_name}! Hope you are having a productive day.",
            f"Welcome, {session.user_name}! Type 'help' to review my available command spectrum."
        ]
        return random.choice(greetings)

    @staticmethod
    def handle_identity(input_text: str, session: UserSession) -> str:
        return (
            f"I am DecodeBot, an enterprise rule-based conversational agent. "
            f"I run on deterministic rule sets without reliance on external neural inference models."
        )

    @staticmethod
    def handle_ai_knowledge(input_text: str, session: UserSession) -> str:
        if "machine learning" in input_text:
            return "Machine Learning (ML) is a subset of AI enabling systems to infer patterns from data automatically."
        elif "data science" in input_text:
            return "Data Science integrates statistics, domain knowledge, and algorithms to extract actionable insights from raw data."
        return (
            "Artificial Intelligence (AI) refers to the computational design of systems capable of mimicking "
            "cognitive functions like reasoning, decision-making, and problem-solving."
        )

    @staticmethod
    def handle_programming_knowledge(input_text: str, session: UserSession) -> str:
        if "why learn python" in input_text:
            return "Python offers high readability, an extensive library ecosystem, and multi-domain utility across AI, cloud, and data engineering."
        elif "programming" in input_text or "coding" in input_text:
            return "Programming is the formal specification of step-by-step logic instructing a machine to complete a given task."
        return "Python is an interpreted, high-level, general-purpose language optimized for clean syntax and rapid development."

    @staticmethod
    def handle_personal_status(input_text: str, session: UserSession) -> str:
        return f"All system diagnostics are optimal, {session.user_name}. Ready for operational requests."

    @staticmethod
    def handle_datetime(input_text: str, session: UserSession) -> str:
        now = datetime.now()
        if "time" in input_text:
            return f"Current System Time: {now.strftime('%H:%M:%S')}"
        return f"Current System Date: {now.strftime('%A, %B %d, %Y')}"

    @staticmethod
    def handle_motivation(input_text: str, session: UserSession) -> str:
        quotes = [
            "\"Quality is not an act, it is a habit.\" — Aristotle",
            "\"Simplicity is prerequisite for reliability.\" — Edsger W. Dijkstra",
            "\"Action is the foundational key to all success.\" — Pablo Picasso",
            "\"Focus on execution over idea generation.\" — Industry Proverb"
        ]
        return random.choice(quotes)

    @staticmethod
    def handle_calculator(input_text: str, session: UserSession) -> str:
        """Parses arithmetic requests in the format: <operation> <num1> <num2>."""
        parts = input_text.split()
        if len(parts) != 3:
            return "Invalid format. Expected execution pattern: add 10 20 | subtract 50 15 | multiply 4 5 | divide 100 4"

        command, raw_a, raw_b = parts[0], parts[1], parts[2]
        try:
            val_a, val_b = float(raw_a), float(raw_b)
            if command == "add":
                return f"Calculation Result: {val_a + val_b:g}"
            elif command == "subtract":
                return f"Calculation Result: {val_a - val_b:g}"
            elif command == "multiply":
                return f"Calculation Result: {val_a * val_b:g}"
            elif command == "divide":
                if val_b == 0:
                    return "Execution Error: Division by zero is undefined."
                return f"Calculation Result: {val_a / val_b:g}"
        except ValueError:
            return "Execution Error: Non-numeric operands provided."
        
        return "Unsupported arithmetic operation."

    @staticmethod
    def handle_help(input_text: str, session: UserSession) -> str:
        return """
+-----------------------------------------------------------------------------------+
|                        DECODEBOT CAPABILITIES & COMMAND SPEC                      |
+-----------------------------------------------------------------------------------+
| 1. Greetings           : hi, hello, hey, assalamualaikum, good morning            |
| 2. Identity Queries    : who are you, what is your name, introduce yourself       |
| 3. AI Domain           : what is ai, what is machine learning, what is data science|
| 4. Software Dev        : what is python, what is programming, why learn python     |
| 5. Status Diagnostics  : how are you, are you fine, how is your day               |
| 6. Temporal Tools      : current time, current date, today's date                 |
| 7. Motivational Core   : inspire me, motivational quote, quote                    |
| 8. Calculator          : add X Y, subtract X Y, multiply X Y, divide X Y          |
| 9. Help Interface      : help                                                     |
| 10. Termination        : exit, quit, bye, goodbye                                 |
+-----------------------------------------------------------------------------------+"""


# ==============================================================================
# INTENT ENGINE
# ==============================================================================

class RuleEngine:
    """Declarative Rule Matching Engine for intent resolution."""

    def __init__(self) -> None:
        self.rules: List[IntentRule] = []
        self._register_default_rules()

    def _register_default_rules(self) -> None:
        """Configures system intent patterns and maps them to business handlers."""
        self.rules = [
            IntentRule(
                intent_id="greeting",
                patterns=["hi", "hello", "hey", "assalamualaikum", "good morning", "good afternoon", "good evening"],
                handler=ResponseHandlers.handle_greeting
            ),
            IntentRule(
                intent_id="identity",
                patterns=["who are you", "what is your name", "introduce yourself", "tell me about yourself"],
                handler=ResponseHandlers.handle_identity
            ),
            IntentRule(
                intent_id="ai_knowledge",
                patterns=["what is ai", "define artificial intelligence", "explain artificial intelligence", "what is machine learning", "what is data science"],
                handler=ResponseHandlers.handle_ai_knowledge
            ),
            IntentRule(
                intent_id="programming_knowledge",
                patterns=["what is python", "what is programming", "why learn python", "what is coding"],
                handler=ResponseHandlers.handle_programming_knowledge
            ),
            IntentRule(
                intent_id="personal_status",
                patterns=["how are you", "are you fine", "how is your day"],
                handler=ResponseHandlers.handle_personal_status
            ),
            IntentRule(
                intent_id="datetime",
                patterns=["current time", "what time is it", "today's date", "current date"],
                handler=ResponseHandlers.handle_datetime
            ),
            IntentRule(
                intent_id="motivation",
                patterns=["motivate me", "give me a quote", "inspiration", "motivational quote", "inspire me"],
                handler=ResponseHandlers.handle_motivation
            ),
            IntentRule(
                intent_id="calculator",
                patterns=[r"^(add|subtract|multiply|divide)\s+\S+\s+\S+"],
                handler=ResponseHandlers.handle_calculator,
                is_regex=True
            ),
            IntentRule(
                intent_id="help",
                patterns=["help", "commands", "menu"],
                handler=ResponseHandlers.handle_help
            )
        ]

    def resolve_response(self, normalized_input: str, session: UserSession) -> str:
        """Evaluates input against registered patterns and returns the mapped handler output."""
        for rule in self.rules:
            if rule.is_regex:
                for pattern in rule.patterns:
                    if re.match(pattern, normalized_input):
                        logger.debug(f"Matched Regex Intent: {rule.intent_id}")
                        return rule.handler(normalized_input, session)
            else:
                if normalized_input in rule.patterns:
                    logger.debug(f"Matched Exact Intent: {rule.intent_id}")
                    return rule.handler(normalized_input, session)

        # Fallback response for unhandled intents
        return "I apologize, but that query falls outside my configured rule base. Type 'help' to view valid options."


# ==============================================================================
# CLI RUNTIME & DRIVER
# ==============================================================================

class ChatbotApplication:
    """Controller class managing input/output streams, normalization, and runtime lifecycle."""

    EXIT_COMMANDS = {"exit", "quit", "bye", "goodbye"}

    def __init__(self) -> None:
        self.engine = RuleEngine()

    def normalize_input(self, text: str) -> str:
        """
        Applies input sanitization:
        1. Lowercases the string.
        2. Strips whitespace.
        3. Removes trailing punctuation/symbols (e.g., '?', '!', '.', ',').
        """
        # Step 1: Lowercase and strip outer whitespace
        text = text.strip().lower()
        
        # Step 2: Remove punctuation characters except letters, digits, and spaces
        cleaned_text = re.sub(r'[^\w\s]', '', text)
        
        # Step 3: Collapse extra spaces into a single space
        return re.sub(r'\s+', ' ', cleaned_text).strip()

    def run(self) -> None:
        """Executes the primary execution loop."""
        print("=====================================================================")
        print("     DECODEBOT CONVERSATIONAL ENGINE (ENTERPRISE EDITION v2.0)       ")
        print("=====================================================================")

        # Session Initialization
        raw_name = input("\nDecodeBot: System initialization complete. Enter user identity: ").strip()
        user_name = raw_name if raw_name else "Engineer"
        session = UserSession(user_name=user_name)

        logger.info(f"Session started for user: {session.user_name}")
        print(f"\nDecodeBot: Welcome, {session.user_name}. Engine initialized. Type 'help' or 'exit'.\n")

        # Main Event Loop
        while True:
            try:
                raw_input = input(f"[{session.user_name}]: ")
                clean_input = self.normalize_input(raw_input)

                # Ignore empty inputs
                if not clean_input:
                    continue

                session.increment_interaction()

                # Check for termination sequence
                if clean_input in self.EXIT_COMMANDS:
                    print(f"\nDecodeBot: Session terminating. Goodbye, {session.user_name}!")
                    print(f"📊 Telemetry Summary: {session.interaction_count} total interaction events recorded.")
                    print("=====================================================================")
                    logger.info(f"Session closed cleanly. Total interactions: {session.interaction_count}")
                    break

                # Resolve and dispatch response
                response = self.engine.resolve_response(clean_input, session)
                print(f"DecodeBot: {response}\n")

            except (KeyboardInterrupt, EOFError):
                print(f"\n\nDecodeBot: Force shutdown signal received. Terminating session...")
                logger.warning("Session interrupted by user (SIGINT/EOF).")
                break


if __name__ == "__main__":
    app = ChatbotApplication()
    app.run()
```
